```python
from pyspark.sql.functions import * 
from pyspark.sql.types import * 
import time 
```


###**Data Reading**

```python
df = spark.read.format("parquet")\
    .load("abfss://bronze@databricklhp.dfs.core.windows.net/customers")
```

```python
df.display()
```

```python
df = df.drop("_rescued_data")
df.display()
```

```python
df = df.withColumn("domains",split(col('email'),'@')[1])
df.display()
```

```python
df.groupBy("domains").agg(count("customer_id").alias("total_customers")).sort("total_customers",ascending=False).display()
```

```python
df_gmail = df.filter(col('domains')=="gmail.com")
df_gmail.display()
time.sleep(5)

df_yahoo = df.filter(col('domains')=="yahoo.com")
df_gmail.display()
time.sleep(5)

df_hotmail = df.filter(col('domains')=="hotmail.com")
df_hotmail.display()
time.sleep(5)
```

```python
df = df.withColumn("full_name",concat(col('first_name'),lit(' '),col('last_name')))
df = df.drop('first_name','last_name')
df.display()
```

```python
df.write.mode("overwrite").format("delta").save("abfss://silver@databricklhp.dfs.core.windows.net/customers")
```
