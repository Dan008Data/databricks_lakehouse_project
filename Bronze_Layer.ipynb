### **Data Reading**  


```python
df = spark.readStream.format("CloudFiles")\
        .option("CloudFiles.format","parquet")\
        .option("CloudFiles.schemalocation",f"abfss://bronze@databricklhp.dfs.core.windows.net/checkpoint_{p_file_name}")\
        .load(f"abfss://source@databricklhp.dfs.core.windows.net/{p_file_name}")
```

```python
df.writeStream.format("parquet")\
        .outputMode("append")\
        .option("checkpointLocation",f"abfss://bronze@databricklhp.dfs.core.windows.net/checkpoint_{p_file_name}")\
        .option("path",f"abfss://bronze@databricklhp.dfs.core.windows.net/{p_file_name}")\
        .trigger(once=True)\
        .start()
```
