```python
from pyspark.sql.functions import *
from pyspark.sql.types import * 
```

```python
init_load_flag = int(dbutils.widgets.get("init_load_flag"))
```

###**Data Reading From Source**

```python
df = spark.sql("select * from databricks_cata.silver.customers_silver")
```

```python
df.display()
```

###**Removing Duplicates**

```python
df = df.dropDuplicates(subset=["customer_id"])
```

###**Dividing New vs Old Records**

```python
if init_load_flag == 0:

    df_old = spark.sql('''select DimCustomerKey, customer_id, create_date, update_date 
                       from databricks_cata.gold.DimCustomers''')
    
else:

    df_old = spark.sql('''select 0 DimCustomerKey, 0 customer_id,0 create_date, 0 update_date
                       from databricks_cata.silver.customers_silver where 1 = 0 ''')
```

```python
df_old.display()
```

###**Renaming Columns of df_old**

```python
df_old = df_old.withColumnRenamed("DimCustomerKey","old_DimCustomerKey")\
    .withColumnRenamed("customer_id","old_customer_id")\
    .withColumnRenamed("create_date","old_create_date")\
    .withColumnRenamed("update_date", "old_update_date")
```

###**Applying Join With the Old Records**

```python
df_join = df.join(df_old, df['customer_id'] == df_old['customer_id'],'left')
```

```python
df_join.display()
```

###**Seperating New vs Old Value**

```python
df_new = df_join.filter(df_join['old_DimCustomerKey'].isNull()) #New records
```

```python
df_old = df_join.filter(df_join['old_DimCustomerKey'].isNotNull()) #Old records 
```

###**Preparing df_old**

```python
#Dropping all the columns which are not required

df_old = df_old.drop('old_customer_id','old_update_date')

#Renaming "old_create_date" , old_DimCustomerKey columns to "create_date"

df_old = df_old.withColumnRenamed('old_create_date','create_date')
df_old = df_old.withColumnRenamed('old_DimCustomerKey','DimCustomerKey')
df_old = df_old.withColumn("create_date",to_timestamp(col("create_date")))

#Recreating "update_date" column with current timestamp 

df_old = df_old.withColumn("update_date",current_timestamp())
```

```python
df_old.display()
```

###**Preparing df_new**

```python
# Dropping all the columns which are not required

df_new = df_new.drop('old_DimCustomerKey','old_customer_id','old_update_date','old_create_date')


# Recreating "update_date", "current_date" columns with current timestamp 
df_new = df_new.withColumn("update_date",current_timestamp())
df_new = df_new.withColumn("create_date",current_timestamp())

```

```python
df_new.display()
```

###**Surrogate Key - On All Values**

```python
df_new = df_new.withColumn("DimCustomerKey",monotonically_increasing_id()+lit(1))
df_new.display()
```

###**Adding Max Surrogate Key**

```python
if init_load_flag ==1: 
    max_surrogate_key = 0
else: 
    df_max_sur = spark.sql("select max(DimCustomerKey) as max_surrogate_key from databricks_cata.gold.DimCustomers") 

    #Converting df_maxsur to max_surrogate_key variable
    max_surrogate_key = df_max_sur.first()["max_surrogate_key"]
    #max_surrogate_key = df.maxsur.collect()[0]["max_surrogate_key"]

```

```python
df_new = df_new.withColumn("DimCustomerKey", lit(max_surrogate_key)+col("DimCustomerKey"))
```

###**Union of df_old and df_new**

```python
df_final = df_new.unionByName(df_old)
```

```python
df_final.display()
```

###**SCD Type-1**

```python
from delta.tables import DeltaTable
```

```python
if spark.catalog.tableExists("databricks_cata.gold.DimCustomers"):
    
    dlt_obj = DeltaTable.forPath(spark, "abfss://gold@databricklhp.dfs.core.windows.net/DimCustomers")

    dlt_obj.alias("trg").merge(df_final.dropDuplicates(["DimCustomerKey"]).alias("src"),"trg.DimCustomerKey = src.DimCustomerKey")\
        .whenMatchedUpdateAll()\
        .whenNotMatchedInsertAll()\
        .execute()
       
else:
     df_final.write.mode("overwrite")\
            .option("path","abfss://gold@databricklhp.dfs.core.windows.net/DimCustomers")\
            .saveAsTable("databricks_cata.gold.DimCustomers")

    
```
